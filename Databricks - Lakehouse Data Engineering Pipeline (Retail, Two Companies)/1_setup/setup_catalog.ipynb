# Catalog Setup

Initializes the `fmcg` Unity Catalog environment used by the rest of the project.

Run this notebook first to create the catalog plus the Bronze, Silver, and Gold schemas.

```python
%sql
-- Create the project catalog if it does not already exist.
CREATE CATALOG IF NOT EXISTS fmcg;
```

```python
%sql
-- Switch the active catalog so subsequent schema commands land in the right workspace.
USE catalog fmcg;
```

```python
%sql
-- Create the medallion-style schemas used throughout the pipeline.
CREATE SCHEMA IF NOT EXISTS fmcg.gold;
CREATE SCHEMA IF NOT EXISTS fmcg.bronze;
CREATE SCHEMA IF NOT EXISTS fmcg.silver;
```

```python
%sql
-- Quick validation check so the setup notebook shows the created schemas.
SHOW DATABASES FROM fmcg;
```
