# Parent Incremental Fact Import

Databricks SQL notebook that loads the parent-company incremental fact extract into the shared Gold fact table before downstream reporting refreshes.


```sql
COPY INTO fmcg.gold.fact_orders
FROM (
  SELECT 
    CAST(date AS DATE) AS date,
    product_code,
    customer_code,
    CAST(sold_quantity AS BIGINT) AS sold_quantity
  FROM '/Volumes/fmcg/gold/parent_incremental_update/fact_orders.csv'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header'='true');
```
